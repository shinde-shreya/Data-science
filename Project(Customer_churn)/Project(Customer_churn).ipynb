# Linear Regression

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
for i in df.columns:
    if df[i].dtype=='object':
        df[i]=le.fit_transform(df[i])
```

```python
df.info()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
x=df[['MonthlyCharges']]
y=df['tenure']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=32)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
y_final_pred=pd.DataFrame(y_pred)
```

```python
y_final_pred
```

```python
from sklearn.metrics import mean_squared_error
```

```python
rmse=np.sqrt(mean_squared_error(y_test,y_final_pred))
rmse
```

```python
errors = y_test - y_pred
df_errors = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred, 'Error': errors})
```

```python
df_errors
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
x=df[['tenure','MonthlyCharges']]
y=df['Churn']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
rfc=RandomForestClassifier(n_estimators=50,random_state=32)
```

```python
rfc.fit(x_train,y_train)
```

```python
y_pred=rfc.predict(x_test)
```

```python
pd.DataFrame({"Actual_value":y_test,"Predicted_value":y_pred})
```

```python
from sklearn.metrics import *
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
accuracy_score(y_test,y_pred)
```
