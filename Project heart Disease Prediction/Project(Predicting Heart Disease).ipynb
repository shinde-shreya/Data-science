# Logistic Regression

```python
x=df.drop(columns=['target'])
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
pd.DataFrame(y_pred)
```

```python
from sklearn.metrics import *
```

```python
logistic_cm=confusion_matrix(y_pred,y_test)
logistic_cm
```

```python
logistic_accuracy=accuracy_score(y_test,y_pred)*100
```

# Decision Tree

```python
x=df.drop(columns=['target'])
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtc=DecisionTreeClassifier(max_depth=5)
```

```python
dtc.fit(x_train,y_train)
```

```python
y_pred=dtc.predict(x_test)
pd.DataFrame(y_pred)
```

```python
decision_cm=confusion_matrix(y_test,y_pred)
decision_cm
```

```python
decision_accuracy=accuracy_score(y_test,y_pred)*100
```

```python
from sklearn import tree
```

```python
Tree=tree.plot_tree(dtc,feature_names=df.columns,filled=True)
```

# Random Forest

```python
x=df.drop(columns=['target'])
y=df['target']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc=RandomForestClassifier(n_estimators=50,random_state=32)
```

```python
rfc.fit(x_train,y_train)
```

```python
y_pred=rfc.predict(x_test)
pd.DataFrame(y_pred)
```

```python
random_cm=confusion_matrix(y_test,y_pred)
random_cm
```

```python
random_accuracy=accuracy_score(y_test,y_pred)*100
```
